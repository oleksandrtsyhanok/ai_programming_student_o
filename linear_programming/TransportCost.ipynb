## Transportkosten

**Doelstelling:**
Minimaliseer de totale transportkosten bij het bepalen van de optimale route van goederen van magazijnen naar klanten.

**Doelstelling Functie:**
$$ \text{Minimaliseer} \quad C = \sum_{i=1}^{N} \sum_{j=1}^{M} c_{ij} \cdot x_{ij}$$

waar:
- $ C $ de totale transportkosten zijn,
- $c_{ij} $ de kosten zijn van het transporteren van goederen van magazijn $i$ naar klant $j$,
- $x_{ij} $ een beslissingsvariabele is die aangeeft of goederen worden getransporteerd van magazijn $i$ naar klant $j$ (1 als wel, 0 indien niet).

**Beperkingen:**
$ \text{Onderhevig aan} \quad \sum_{i=1}^{N} x_{ij} = 1 \quad \text{voor elke klant } j $, 
$ \sum_{j=1}^{M} x_{ij} \leq 1 \quad \text{voor elk magazijn } i $, 
$ x_{ij} \in \{0, 1\} \quad \text{binaire beperkingen} $

Het doel is om de totale transportkosten te minimaliseren door elke klant toe te wijzen aan exact één magazijn en ervoor te zorgen dat elk magazijn hoogstens één klant bedient.
